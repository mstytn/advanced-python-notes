# Context Managers

for using `with` `as` context.


```python
class File:
    def __init__(self, filename):
        self.filename = filename
    
    def __enter__(self):
        print(f'Opening {self.filename}')
        return self
    
    def __exit__(self, exec_type, exc_val, exc_tb):
        print(f'Closing {self.filename}')
        return False


with File('do.txt') as f:
    print(f.filename)

print('done')

```
