# Async IO


```python
import asyncio


async def fetch_data(data: int)-> dict:
    print('Fetching Data...')
    await asyncio.sleep(2)
    return {'data': data}

async def counter():
    for i in range(10):
        await asyncio.sleep(.5)
        print(i)

# Throws two tasks in the background at the same time
async def main() -> str:
    task1 = asyncio.create_task(fetch_data(10))
    task2 = asyncio.create_task(counter())

    data: dict = await task1
    print(data)
    await task2
    return 'done'


#asyncio.run(main())
await main()
```

```python
async def main() -> None:
    # Automatically schedules and awaits all tasks when exiting the 'async with' block
    async with asyncio.TaskGroup() as tg:
        task1 = tg.create_task(fetch_data(10))
        task2 = tg.create_task(counter())

    # Once outside the block, both tasks are guaranteed to be finished
    print(task1.result(), task2.result()) # Still fires the tasks in parallel but waits for both completed before returning

await main()
```

## Tasks

### Cancel

```python
async def fetch_data_10(data: int)-> dict:
    print('Fetching Data...')
    await asyncio.sleep(10)
    return {'data': data}

async def main():
    task = asyncio.create_task(fetch_data_10(100))
    await asyncio.sleep(5)

    task.cancel()
    await asyncio.sleep(.1)
    print(task.cancelled())

# Alternative 

    try:
        data: dict = await task
        print(data)
    except asyncio.CancelledError:
        print('Task was cancelled')

    try:
        if task.done():
            data: dict = task.result()
            print(data)
        else:
            print('Data not Ready')
    except asyncio.CancelledError:
        print('Task was cancelled')
    

await main()
```

### Timeout


```python
task = asyncio.create_task(fetch_data_10(100))
await asyncio.wait_for(task, timeout=5)
```
