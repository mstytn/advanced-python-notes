# Data Classes


```python
class Fruit:
    def __init__(self, name: str, cal: float):
        self.name = name
        self.calories = cal

    def __eq__(self, other):
        return self.__dict__ == other.__dict__

apple: Fruit = Fruit('Apple', 10)
apple2: Fruit = Fruit('Apple', 9)

print('Same' if apple == apple2 else 'Different')
```

```python
from dataclasses import dataclass


@dataclass
class DFruit:
    name: str
    calories: float

apple: DFruit = DFruit('Apple', 10)
apple2: DFruit = DFruit('Apple', 9)

print('Same' if apple == apple2 else 'Different')
```
