# 11 - Built-in Functions


## 1. Built-ins Overview (`max()`, `min()`, `sum()`)

Python yorumlayıcısına doğrudan gömülü olarak gelen ve herhangi bir `import` işlemi gerektirmeden her kapsamda erişilebilen yerleşik fonksiyonlardır.

**Syntax**
```python
max(iterable, *, key=None, default=...)
min(iterable, *, key=None, default=...)
sum(iterable, /, start=0)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Yinelenebilir (iterable) veri yapıları üzerinde en büyük/en küçük değeri bulmak veya sayısal toplam almak gibi temel işlemleri C hızında ve tek satırda gerçekleştirmek için kullanılır.
* **Dikkat Edilmesi Gerekenler:** `max()` ve `min()` boş bir iterable aldığında `ValueError` fırlatır; bunu önlemek için `default` parametresi verilmelidir. Farklı tipleri (örneğin `int` ve `str`) aynı koleksiyonda karşılaştırmak `TypeError` üretir.


````python
latencies_ms: list[int] = [42, 18, 95, 31, 64]
empty_samples: list[int] = []

print(max(latencies_ms))
print(min(latencies_ms))
print(sum(latencies_ms))
print(max(empty_samples, default=0))
````

## 2. `print()`

Verilen nesneleri metne dönüştürerek standart çıktıya (`sys.stdout`) veya belirtilen bir dosya akışına yazdıran yerleşik fonksiyondur.

**Syntax**
```python
print(*objects, sep=' ', end='\n', file=None, flush=False)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Konsola biçimlendirilmiş çıktı basmak, birden fazla değeri özel ayraçlarla (`sep`) birleştirmek veya satır sonu davranışını (`end`) değiştirmek için kullanılır.
* **Dikkat Edilmesi Gerekenler:** `sep`, `end`, `file` ve `flush` mutlaka anahtar kelime argümanı (keyword-only) olarak verilmelidir; aksi halde yazdırılacak nesne olarak algılanır. Uzun süren döngülerde çıktının tamponda beklemeden anında ekrana düşmesi için `flush=True` kullanılır.


````python
print('192', '168', '1', '1', sep='.')
print('Initializing modules', end='... ')
print('Done', flush=True)
````

## 3. `enumerate()`

Bir iterable nesne üzerinde dönerken her elemanı otomatik bir sayaç değeriyle eşleştirip `(index, item)` demetleri üreten tembel (lazy) yineleyicidir.

**Syntax**
```python
enumerate(iterable, start=0)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Döngü içinde hem elemanın kendisine hem de sıra numarasına ihtiyaç duyulduğunda `range(len(...))` kalıbı yerine kullanılır.
* **Dikkat Edilmesi Gerekenler:** Kullanıcıya dönük numaralandırmalarda `start=1` parametresi kullanılabilir; ancak bu yalnızca üretilen sayaç değerini kaydırır, asıl koleksiyonun indekslemesini değiştirmez.


````python
audio_stems: list[str] = ['drums.wav', 'bass.wav', 'guitars.wav', 'vocals.wav']

for track_no, stem in enumerate(audio_stems, start=1):
    print(f'{track_no:02d} -> {stem}')
````

## 4. `round()`

Sayısal bir değeri belirtilen ondalık basamak sayısına göre yuvarlayan yerleşik fonksiyondur.

**Syntax**
```python
round(number, ndigits=None)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Ondalıklı sonuçları belirli bir basamak hassasiyetine yuvarlamak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** Python tam ortada kalan (`.5`) değerlerde klasik yukarı yuvarlama yerine **Banker's Rounding** (en yakın çift sayıya yuvarlama: `round(2.5) == 2`, `round(3.5) == 4`) uygular. Ayrıca ikili kayan nokta (IEEE 754) temsili nedeniyle `round(2.675, 2)` gibi ifadeler `2.67` verebilir; finansal hesaplamalarda `decimal.Decimal` kullanılmalıdır.


````python
ratio: float = 10 / 6

print(round(ratio))
print(round(ratio, 2))
print(round(2.5), round(3.5))
print(round(14850, -2))
````

## 5. `range()`

Belirli bir başlangıç, bitiş ve adım değerine göre sabit bellek alanıyla ($O(1)$) tam sayı dizisi temsil eden değişmez (immutable) dizi tipidir.

**Syntax**
```python
range(stop)
range(start, stop, step=1)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Belirli sayıda tekrar eden döngüler kurmak veya tüm sayıları belleğe yüklemeden büyük sayı aralıkları üzerinde üyelik (`in`) ve indeks sorgusu yapmak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** `stop` değeri aralığa dahil değildir (yarı açık aralık: `[start, stop)`). Geriye doğru sayım yapılırken `step` mutlaka negatif verilmelidir, aksi halde boş aralık döner.


````python
import sys

huge_range: range = range(0, 10**20, 5)
countdown: list[int] = list(range(10, 0, -2))

print(sys.getsizeof(huge_range))
print(500_000 in huge_range)
print(countdown)
````

## 6. `globals()`

İçinde bulunulan modülün küresel (global) isim alanındaki tüm değişken, fonksiyon, sınıf ve modül tanımlamalarını içeren canlı sözlüğü (`dict`) döndürür.

**Syntax**
```python
globals() -> dict[str, Any]
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Modül düzeyindeki sembolleri çalışma zamanında incelemek (introspection) veya dinamik adlarla global tanımlara erişmek için kullanılır.
* **Dikkat Edilmesi Gerekenler:** Döndürülen sözlük doğrudan modülün gerçek isim alanına bağlıdır; bu sözlük üzerinde yapılan ekleme, silme veya güncellemeler modül değişkenlerini anında değiştirir.


````python
APP_ENV: str = 'production'

def health_check() -> str:
    return 'OK'

global_symbols: dict[str, object] = globals()
print(global_symbols['APP_ENV'])
print('health_check' in global_symbols)
````

## 7. `locals()`

Bulunulan yerel kapsamın (fonksiyon veya sınıf gövdesi) yerel sembol tablosunu temsil eden bir sözlük döndürür.

**Syntax**
```python
locals() -> dict[str, Any]
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Hata ayıklama sırasında bir fonksiyon içindeki tüm yerel değişkenlerin anlık değerlerini tek seferde görmek veya şablon/format dizgilerine yerel değişkenleri aktarmak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** Modül düzeyinde çağrıldığında `globals()` ile aynı sözlüğü döndürür. Bir fonksiyon içinde ise yerel kapsamın bir anlık kopyasını döndürür; döndürülen sözlüğü değiştirmek fonksiyon içindeki yerel değişkenleri güncellemez.


````python
def inspect_scope(host: str, port: int) -> None:
    use_ssl: bool = True
    local_vars: dict[str, object] = locals()
    print(local_vars)


inspect_scope('10.0.0.12', 5432)
````

## 8. `all()` & `any()`

Bir iterable içindeki elemanların doğruluk (truthiness) değerlerini kısa devre (short-circuit) mantığıyla denetleyen yerleşik fonksiyonlardır: `all()` tüm elemanlar `True` ise, `any()` ise en az bir eleman `True` ise `True` döndürür.

**Syntax**
```python
all(iterable) -> bool
any(iterable) -> bool
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Birden fazla koşulu veya bir koleksiyonun tamamını `for` döngüsü kurmadan tek ifadede doğrulamak için generator ifadeleriyle birlikte kullanılır.
* **Dikkat Edilmesi Gerekenler:** Kısa devre çalıştıkları için ilk belirleyici elemanda (`all` için `False`, `any` için `True`) dururlar. Boş iterable verildiğinde mantıksal boşluk (vacuous truth) gereği `all([])` `True`, `any([])` ise `False` döndürür.


````python
service_checks: dict[str, bool] = {
    'database': True,
    'redis_cache': True,
    'worker_queue': False,
}

all_healthy: bool = all(service_checks.values())
any_healthy: bool = any(service_checks.values())
has_high_load: bool = any(cpu > 85 for cpu in [22, 47, 91, 30])

print(all_healthy, any_healthy, has_high_load)
````

## 9. `isinstance()`

Bir nesnenin belirtilen sınıfın veya demet halinde verilen sınıflardan herhangi birinin örneği (ya da alt sınıfı) olup olmadığını kontrol eden yerleşik fonksiyondur.

**Syntax**
```python
isinstance(object, classinfo) -> bool
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Çalışma zamanında tip doğrulaması yapmak ve kalıtım hiyerarşilerini gözeterek güvenli tip daraltma (type narrowing) sağlamak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** Python'da `bool`, `int` sınıfının alt sınıfıdır; dolayısıyla `isinstance(True, int)` sonucu `True` döner. Python 3.10 ve sonrasında `classinfo` parametresine demet yerine doğrudan birleşim tipi (`int | float`) verilebilir.


````python
class Device:
    pass

class Sensor(Device):
    pass

temp_sensor: Sensor = Sensor()
payload: int | str = 404

print(isinstance(temp_sensor, Sensor))
print(isinstance(temp_sensor, Device))
print(isinstance(payload, int | float))
````

## 10. `callable()`

Bir nesnenin fonksiyon gibi parantez `()` ile çağrılabilir olup olmadığını kontrol eden yerleşik fonksiyondur.

**Syntax**
```python
callable(object) -> bool
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Dinamik olarak alınan bir nesneyi veya geri çağırma (callback) parametresini çalıştırmadan önce `TypeError: '...' object is not callable` hatasını önlemek için kullanılır.
* **Dikkat Edilmesi Gerekenler:** Fonksiyonlar, metotlar, sınıflar ve `__call__()` dunder metodunu tanımlayan sınıflardan türetilen nesneler için `True` döndürür.


````python
class GainProcessor:
    def __call__(self, sample: float) -> float:
        return sample * 1.5


def normalize(val: float) -> float:
    return val


processor: GainProcessor = GainProcessor()
label: str = 'Master'

print(callable(normalize))
print(callable(GainProcessor))
print(callable(processor))
print(callable(label))
````

## 11. `filter()`

Bir iterable içindeki her elemanı verilen doğrulama fonksiyonundan geçirerek yalnızca `True` sonucu üreten elemanları döndüren tembel (lazy) yineleyicidir.

**Syntax**
```python
filter(function, iterable)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Mevcut bir doğrulama fonksiyonunu (predicate) koleksiyon elemanlarına tembel değerlendirme (lazy evaluation) ile uygulamak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** `function` argümanı olarak `None` verilirse koleksiyondaki tüm falsy değerler (`0`, `''`, `None`, `False`) elenir. Sonuç bir `filter` yineleyicisidir; tekrar kullanılabilir bir koleksiyon gerekiyorsa `list()` veya `tuple()` ile tüketilmelidir.


````python
mixed_records: list[str | int | None] = ['node-1', 200, '', 'node-2', None, 503]

def is_valid_host(item: object) -> bool:
    return isinstance(item, str) and len(item) > 0

hosts: list[str] = list(filter(is_valid_host, mixed_records))
truthy_only: list[str | int | None] = list(filter(None, mixed_records))

print(hosts)
print(truthy_only)
````

## 12. `map()`

Verilen dönüştürme fonksiyonunu bir veya daha fazla iterable'ın her elemanına sırayla uygulayarak sonuçları tembel (lazy) olarak üreten yineleyicidir.

**Syntax**
```python
map(function, iterable, *iterables)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Koleksiyon elemanlarına hazır bir fonksiyonu (`str.strip`, `int`, özel dönüşüm fonksiyonu) bellek dostu şekilde uygulamak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** Birden fazla iterable verildiğinde `map()`, en kısa iterable tükendiğinde durur. Tıpkı `filter()` gibi tek kullanımlık bir yineleyici döndürür.


````python
raw_ports: list[str] = [' 80 ', '443', ' 8080']
cleaned_ports: list[int] = list(map(int, map(str.strip, raw_ports)))

base_vals: list[int] = [2, 3, 4]
exponents: list[int] = [3, 2, 2]
powers: list[int] = list(map(pow, base_vals, exponents))

print(cleaned_ports)
print(powers)
````

## 13. `sorted()`

Herhangi bir iterable yapının elemanlarını sıralayarak orijinal veriye dokunmadan **yeni bir sıralı liste** döndüren yerleşik fonksiyondur.

**Syntax**
```python
sorted(iterable, *, key=None, reverse=False) -> list
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Orijinal koleksiyonu bozmamak gerektiğinde veya `list.sort()` metoduna sahip olmayan (`tuple`, `set`, `dict`, generator) yapıları sıralamak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** `list.sort()` listeyi yerinde (in-place) değiştirir ve `None` döndürürken, `sorted()` daima yeni bir `list` döndürür. Özel nesneleri veya büyük/küçük harf duyarsız metinleri sıralamak için `key` parametresine karşılaştırma anahtarı üreten fonksiyon verilmelidir.


````python
tags: list[str] = ['v2', 'Alpha', 'beta', 'Release']
case_insensitive: list[str] = sorted(tags, key=str.casefold)

files: list[tuple[str, int]] = [('intro.mp4', 45), ('outro.mp4', 12), ('main.mp4', 320)]
by_size_desc: list[tuple[str, int]] = sorted(files, key=lambda item: item[1], reverse=True)

print(case_insensitive)
print(by_size_desc)
````

## 14. `eval()`

Metin (`str`) veya derlenmiş kod nesnesi olarak verilen tek bir Python **ifadesini (expression)** değerlendirip sonucunu döndüren yerleşik fonksiyondur.

**Syntax**
```python
eval(expression, globals=None, locals=None)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Yalnızca güvenilir kaynaktan gelen dinamik ifadelerin hesaplanmasında kullanılır.
* **Dikkat Edilmesi Gerekenler:** Atama (`x = 5`), döngü (`for`) veya koşul blokları (`if`) gibi deyimleri (statements) çalıştıramaz; `SyntaxError` verir. Dış kullanıcıdan gelen girdilerde kullanılması kritik güvenlik açığına (Remote Code Execution) yol açar. Yalnızca veri literallerini ayrıştırmak için `ast.literal_eval()` tercih edilmelidir.


````python
formula: str = '(base_rate * hours) + bonus'
context: dict[str, int] = {'base_rate': 150, 'hours': 8, 'bonus': 200}

total_pay: int = eval(formula, {'__builtins__': {}}, context)
print(total_pay)
````

## 15. `exec()`

Çok satırlı kod blokları, fonksiyon/sınıf tanımları, döngüler ve atamalar dahil her türlü dinamik Python **deyimini (statement)** çalıştıran yerleşik fonksiyondur.

**Syntax**
```python
exec(object, globals=None, locals=None, /, *, closure=None) -> None
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Eklenti sistemleri, konfigürasyon betikleri veya çalışma zamanında dinamik kod üretimi gereken özel senaryolarda kullanılır.
* **Dikkat Edilmesi Gerekenler:** `eval()`'dan farklı olarak hiçbir zaman değer döndürmez (`None` döner); üretilen değişkenlere erişmek için bir `locals` sözlüğü geçilmelidir. Doğrulanmamış kullanıcı girdileriyle kullanılması sistemin tamamen ele geçirilmesine neden olabilir.


````python
script_block: str = '''
results = [n ** 2 for n in range(1, 5)]
total = sum(results)
'''

namespace: dict[str, object] = {}
exec(script_block, {'__builtins__': {'range': range, 'sum': sum}}, namespace)

print(namespace['results'])
print(namespace['total'])
````

## 16. `zip()`

Birden fazla iterable'ın aynı sıradaki elemanlarını paralel olarak birleştirip demetler (`tuple`) üreten tembel (lazy) yineleyicidir.

**Syntax**
```python
zip(*iterables, strict=False)
```

**Kullanım Alanları ve Dikkat Edilmesi Gerekenler**
* **Ne Zaman Kullanılır:** Paralel listeler/demetler üzerinde aynı anda gezinmek veya anahtar-değer listelerinden `dict(zip(keys, values))` ile sözlük oluşturmak için kullanılır.
* **Dikkat Edilmesi Gerekenler:** Varsayılan olarak (`strict=False`) en kısa iterable bittiğinde sessizce durur ve fazla elemanları göz ardı eder. Python 3.10 ile eklenen `strict=True` parametresi, girdi uzunlukları eşit değilse sessiz veri kaybını önlemek için `ValueError` fırlatır.


````python
channels: tuple[str, ...] = ('CH1', 'CH2', 'CH3')
instruments: tuple[str, ...] = ('Kick', 'Snare', 'Hi-Hat')
pans: tuple[int, ...] = (0, -15, 25)

routing: dict[str, str] = dict(zip(channels, instruments, strict=True))
print(routing)

for ch, inst, pan in zip(channels, instruments, pans, strict=True):
    print(f'{ch}: {inst} (Pan: {pan})')
````
