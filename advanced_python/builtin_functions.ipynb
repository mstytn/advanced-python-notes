# Builtin Functions of PYTHON

## All & Any

### All

The all() function in Python is a built-in function that checks if all elements
in an iterable (like a list, tuple, set, or string) are true.
If the iterable is empty, it returns True. If any element is False, it returns False.


```python
is_connected: bool = True
has_electricity: bool = True
has_paid: bool = True

has_internet: bool = all([is_connected, has_electricity, has_paid])

print ('Internet' if has_internet else 'Not Connected')
```

### Any

The any() function in Python is a built-in function that checks if at least one element in an iterable is true. If the iterable is empty, it returns False. If any element is True, it returns True.

- Parameters
**iterable:** An iterable object such as a list, tuple, set, or string.
- Return Value
**True:** If at least one element in the iterable is true.
**False:** If all elements in the iterable are false.
**False:** If the iterable is empty.


## Class Checks

### isinstance

The isinstance() function in Python is a built-in function that checks if an object is an instance of a specified class or a tuple of classes. It returns True if the object is an instance of the class or any class in the tuple; otherwise, it returns False.

- Parameters
**object:** The object to check.
**classinfo:** A class or a tuple of classes.
- Return Value
**True:** If the object is an instance of the specified class or any class in the tuple.
**False:** If the object is not an instance of the specified class or any class in the tuple.


```python
class Fruit:
	def __init__(self, name: str):
		self.name = name
		
apple: Fruit = Fruit('Apple')
banana: Fruit =  Fruit('Banana')
	
print(isinstance(apple, Fruit))
print(isinstance(banana, Fruit))
print(isinstance('Banana', str))
```

### callable

The `callable()` function in Python is a built-in function that checks if an object appears callable. In other words, it returns True if the object is callable (i.e., it can be invoked as a function) and False otherwise.

- Parameters
**object:** The object to check.
- Return Value
**True:** If the object is callable.
**False:** If the object is not callable.


```python
a: str = 'a'

def do_something():
	pass

def b():
	pass
	
if __name__ == '__main__':
	print(callable(a)) # False
	print(callable(do_something)) # True
	print(callable(b)) # True
```

## Lists

### filter


```python
people: list[str, int] = ['Mario', 'Luigi', 10, 'Toad', 20]

def is_str(element: object) -> bool:
	return isinstance(element, str)

filtered_people: list[str] = list(filter(is_str, people))
print(filtered_people)
```

### map


```python
numbers: list[int] = [1, 2, 3, 4, 5, 6]

def convert_to_str(element: object) -> str:
	return str(element)
	
converted_list: list[str] = list(map(convert_to_str, numbers))
print(converted_list)
```

### sorted


```python
from random import shuffle

numbers: list[int] = [1, 2, 3, 4, 5, 6]
numbers.sort() # easy sort | sorts the list in place 
shuffle(numbers)

sorted_nums: list[int] = sorted(numbers) # reverse: bool
print(sorted_nums)
```

### Manual Sorting

Sorting with a custom callable

The sorted() function in Python is a built-in function that returns a new sorted list from the elements of any iterable. It can also sort objects with named attributes.

- Parameters

**iterable:** The iterable to sort.

**key (***optional***):** A function that takes a single argument and returns a key to use for sorting. By default, it is None, meaning the elements themselves are sorted.

**reverse (***optional***):** A boolean indicating whether to sort the list in descending order (True) or ascending order (False). The default is False.

- Summary

The sorted() function is a versatile tool for sorting iterables. When dealing with objects with named attributes, you can use a lambda function as the key parameter to sort the objects based on the value of their named attributes.


```python
class Fruit:
    def __init__(self, name: str, calories: float):
        self.name = name
        self.calories = calories
    
    def __repr__(self):
        return f'{self.name}: {self.calories}'

# Corrected list definition
f: list[Fruit] = [
    Fruit('Apple', 58),
    Fruit('Banana', 150),
    Fruit('Pineapple', 100),
]

def sort_by_calories(fruit: Fruit) -> float:
    return fruit.calories

# Corrected sorted function call
sorted_fruits: list[Fruit] = sorted(f, key=sort_by_calories)
print(sorted_fruits)
```

## Eval vs Exec

The `eval()` and `exec()` functions in Python are both used to execute Python code dynamically, but they serve different purposes and have different behaviors.

### `eval()`

- **Purpose**: The `eval()` function evaluates a string as a Python expression and returns the result.
- **Syntax**: `eval(expression, globals=None, locals=None)`
- **Parameters**:
  - `expression`: A string representing a Python expression.
  - `globals` (optional): A dictionary or mapping object that defines the global namespace for the evaluation.
  - `locals` (optional): A dictionary or mapping object that defines the local namespace for the evaluation.
- **Return Value**: The result of the evaluated expression.
- **Security Consideration**: `eval()` can execute arbitrary code, which can be a security risk if the input is not trusted. It should be used carefully and with caution.

### `exec()`

- **Purpose**: The `exec()` function executes Python code from a string or a code object.
- **Syntax**: `exec(object, globals=None, locals=None)`
- **Parameters**:
  - `object`: A string containing Python code or a code object.
  - `globals` (optional): A dictionary or mapping object that defines the global namespace for the execution.
  - `locals` (optional): A dictionary or mapping object that defines the local namespace for the execution.
- **Return Value**: `None`
- **Security Consideration**: Similar to `eval()`, `exec()` can execute arbitrary code, which can be a security risk if the input is not trusted. It should be used carefully and with caution.

### Example Usage


```python
# Evaluating a simple expression
result = eval('2 + 3 * 4')
print(result)  # Output: 14

# Using globals and locals
x = 1
y = 2
result = eval('x + y', {}, {'x': 10, 'y': 20})
print(result)  # Output: 30
```

### Summary

- **`eval()`**: Evaluates a Python expression and returns the result.

## `zip()`

The `zip()` function in Python is a built-in function that takes multiple iterables and returns an iterator of tuples, where the i-th tuple contains the i-th element from each of the input iterables. If the iterables are of different lengths, the `zip()` function stops when the shortest iterable is exhausted.

### Parameters

- `iterables`: One or more iterable objects (e.g., lists, tuples, strings, etc.).

### Return Value

- An iterator of tuples, where each tuple contains elements from the corresponding iterables.

### Example Usage


```python
# Example 1: Zipping two lists
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
zipped_data = zip(names, ages)
print(list(zipped_data))  # Output: [('Alice', 25), ('Bob', 30), ('Charlie', 35)]

# Example 2: Zipping three lists
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
genders = ['Female', 'Male', 'Male']
zipped_data = zip(names, ages, genders)
print(list(zipped_data))  # Output: [('Alice', 25, 'Female'), ('Bob', 30, 'Male'), ('Charlie', 35, 'Male')]

# Example 3: Zipping with different lengths
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30]
zipped_data = zip(names, ages)
print(list(zipped_data))  # Output: [('Alice', 25), ('Bob', 30)]

# Example 4: Unzipping
zipped_data = [('Alice', 25), ('Bob', 30), ('Charlie', 35)]
names, ages = zip(*zipped_data)
print(names)  # Output: ('Alice', 'Bob', 'Charlie')
print(ages)   # Output: (25, 30, 35)
```

### Summary

The `zip()` function is useful for combining multiple iterables into a single iterator of tuples, making it easy to iterate over them in parallel. It is often used in data processing and analysis tasks where elements from multiple sources need to be combined.
