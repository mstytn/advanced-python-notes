# Generators

## Yielding values


```python
def gen(n: int):
    for i in range(n):  # noqa: UP028
        yield i


generator = gen(3)
print(next(generator))  # 0
print(next(generator))  # 1
print(next(generator))
```

```python
def gen_more(n: int):
    yield from range(n)

generator = gen(10**10)

list_a: list[int] = []
for number in generator:
    if number == 1000:
        break
    list_a.append(number)

print (len(list_a))  # 1000
```

### Looping the generators


```python
yield_comprehension = (i for i in range(10))

while (t := next(yield_comprehension, None)) is not None:
    print(t)

# for t in yield_comprehension:
#     print(t)
```
